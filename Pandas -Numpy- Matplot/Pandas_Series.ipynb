### Example 1: Creating Series Using List

- Index

    - Fully customize

```python
import pandas as pd


lst = [10, 20, 30, 40, 50, 60]
s = pd.Series(lst)
print(s, type(s))
```

```python
lst = [10, 20, 30, 40, 50, 60]
s = pd.Series(lst, dtype=float)
print(s, type(s))
```

# Common Operation with Series

```python
import pandas as pd

# Creating a Pandas Series
s = pd.Series([10, 20, 30, 40, 50, None, 20])

print("Original Series:")
print(s)


# =========================
# 1. ARITHMETIC OPERATIONS
# =========================

print("\nAdd 10 to every value:")
print(s + 10)          # Adds 10 to each value

print("\nMultiply every value by 2:")
print(s * 2)           # Multiplies each value by 2


# =========================
# 2. AGGREGATION OPERATIONS
# =========================

print("\nSum:", s.sum())        # Total of all values
print("Mean:", s.mean())        # Average value
print("Minimum:", s.min())      # Smallest value
print("Maximum:", s.max())      # Largest value
print("Count:", s.count())      # Counts non-null values


# =========================
# 3. COMPARISON OPERATIONS
# =========================

print("\nValues greater than 25:")
print(s > 25)          # Returns True or False for each value


# =========================
# 4. FILTERING
# =========================

print("\nValues greater than 25:")
print(s[s > 25])       # Selects only values greater than 25


# =========================
# 5. ACCESSING ELEMENTS
# =========================

print("\nFirst element:")
print(s.iloc[0])       # Access value using position

print("\nThird element:")
print(s.iloc[2])       # Access value at position 2


# =========================
# 6. SORTING
# =========================

print("\nSorted values:")
print(s.sort_values())  # Sorts values in ascending order


# =========================
# 7. MISSING VALUES
# =========================

print("\nCheck missing values:")
print(s.isna())         # Returns True where value is missing

print("\nNumber of missing values:")
print(s.isna().sum())   # Counts total missing values

print("\nRemove missing values:")
print(s.dropna())       # Removes None/NaN values

print("\nReplace missing values with 0:")
print(s.fillna(0))      # Replaces missing values with 0


# =========================
# 8. UNIQUE VALUES
# =========================

print("\nUnique values:")
print(s.unique())       # Returns unique values

print("\nNumber of unique values:")
print(s.nunique())      # Counts unique values


# =========================
# 9. VALUE COUNTS
# =========================

print("\nFrequency of each value:")
print(s.value_counts()) # Counts how many times each value appears


# =========================
# 10. APPLY FUNCTION
# =========================

print("\nSquare of each value:")
print(s.apply(lambda x: x ** 2))  # Applies square operation to every value


# =========================
# 11. STRING OPERATIONS
# =========================

# Creating a Series with strings
names = pd.Series(["mohit", "rahul", "aman"])

print("\nOriginal names:")
print(names)

print("\nConvert to uppercase:")
print(names.str.upper())       # Converts all text to uppercase

print("\nConvert to lowercase:")
print(names.str.lower())       # Converts all text to lowercase

print("\nCheck if name contains 'a':")
print(names.str.contains("a")) # Returns True if 'a' is present
```

# Vectorized Operations in Pandas Series

```python
import pandas as pd

# Creating a Series
s = pd.Series([10, 20, 30, 40, 50])

print("Original Series:")
print(s)


# ==============================
# VECTOR ARITHMETIC OPERATIONS
# ==============================

print("\nAdd 10 to every element:")
print(s + 10)          # Adds 10 to each value

print("\nSubtract 5 from every element:")
print(s - 5)           # Subtracts 5 from each value

print("\nMultiply every element by 2:")
print(s * 2)           # Multiplies each value by 2

print("\nDivide every element by 2:")
print(s / 2)           # Divides each value by 2

print("\nSquare every element:")
print(s ** 2)          # Raises every value to power 2


# ==============================
# VECTOR COMPARISON OPERATIONS
# ==============================

print("\nCheck values greater than 25:")
print(s > 25)          # Compares every value with 25


# ==============================
# VECTOR OPERATION BETWEEN
# TWO SERIES
# ==============================

s1 = pd.Series([10, 20, 30])
s2 = pd.Series([1, 2, 3])

print("\nAddition of two Series:")
print(s1 + s2)         # [11, 22, 33]

print("\nMultiplication of two Series:")
print(s1 * s2)         # [10, 40, 90]
```

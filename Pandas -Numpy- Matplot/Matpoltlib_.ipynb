# BarChart

```python
import matplotlib.pyplot as plt

Subject=["Python","C","Java","DSA"]
Score=[40,60,65,85]


plt.bar(Subject,Score, color=["red", "Blue", "Green","skyblue"])

plt.title("Marks of student")
plt.xlabel("Score")

plt.ylabel("Marks")

plt.grid(axis='y')

plt.show()
```

# Pie Chart

- used for data distribution
- tell us 100% data which lies in which categories

- a single part of pie chart is called slice
- name given to slice is know as label



```python
subject=["English","GK", "Apptitude", "Technical"]

score=[50,40,20,12]


plt.pie(score,labels=subject,autopct="%1.1f%%")

plt.title("Neet Exam Marks Distribution")

plt.show()
```

```python
subject=["English","GK", "Apptitude", "Technical"]

score=[50,40,20,12]


plt.pie(score, startangle=180, labels=subject,autopct="%1.1f%%")

plt.title("Neet Exam Marks Distribution")

plt.show()
```

```python
subject=["English","GK", "Apptitude", "Technical"]

score=[50,40,20,12]


plt.pie(score, startangle=180, explode=(0,0,0,0.5), labels=subject,autopct="%1.1f%%")

plt.title("Neet Exam Marks Distribution")

plt.show()
```

```python
subject=["English","GK", "Apptitude", "Technical"]

score=[50,40,20,12]
color=["red","blue","green","yellow"]


plt.pie(score, startangle=180, explode=(0,0,0,0.5), labels=subject,autopct="%1.1f%%", colors=color)

plt.title("Neet Exam Marks Distribution")

plt.show()
```

```python
subject=["English","GK", "Apptitude", "Technical"]

score=[50,40,20,12]
color=["red","blue","green","yellow"]


plt.pie(score, startangle=180, explode=(0,0,0,0.5), labels=subject,autopct="%1.1f%%", colors=color,shadow=True)

plt.title("Neet Exam Marks Distribution")

plt.show()
```

```python
from matplotlib import legend
subject=["English","GK", "Apptitude", "Technical"]

score=[50,40,20,12]
color=["red","blue","green","yellow"]


plt.pie(score, startangle=180, explode=(0,0,0,0.5), labels=subject,autopct="%1.1f%%", colors=color,shadow=True)



plt.title("Neet Exam Marks Distribution")

plt.legend(loc="lower left")                 # Adjust Legend

plt.show()
```

# Scatter Plot

- Find realtion it may be (Postive , Negative or Nutural)

- Postive relation (age )

- negative one increase second decrease
    - ex : you are a biker and need to reach to point A to B as you cover distance but time is decrease

- Nutrual relation both same


- It find realtion and co-realtion and help to identify to outlier

```python
import matplotlib.pyplot as plt

x=[1,5,10,50]
y=[10,20,30,100]

plt.scatter(x,y)

plt.title("Scatter Graph")

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,5,10,50]
y=[10,20,30,100]
color=["red","blue","green","yellow"]
size=[100,250,500,800]

plt.scatter(x,y,c=color,s=size)


plt.title("Scatter Graph")

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,5,10,50]
y=[10,20,30,100]
color=["red","blue","green","yellow"]
size=[100,250,500,800]

plt.scatter(x,y,c=color,s=size, alpha=0.7)


plt.title("Scatter Graph")

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,5,10,50]
y=[10,20,30,100]
color=["red","blue","green","yellow"]
size=[100,250,500,800]

plt.scatter(x,y,c=color,s=size, alpha=0.7)


plt.title("Scatter Graph")

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.colorbar(label="Intensity")

plt.show()
```

```python
import matplotlib.pyplot as plt

x1=[1,5,10,50]
y1=[10,20,30,100]

x2=[1,5,10,50]
y2=[10,20,30,100]


color=["red","blue","green","yellow"]
size=[100,250,500,800]

plt.scatter(x1,y1,c=color,s=size, alpha=0.7)
plt.scatter(x2,y2,c=color,s=size, alpha=0.7)


plt.title("Scatter Graph")

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.colorbar(label="Intensity")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o")

for i in range(len(x)):
    plt.text(x[i], y[i], str(y[i]))

plt.show()
```

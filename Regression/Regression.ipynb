```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    'Area': [1000, 1500, 2000, 2500, 1800, 2200],
    'Bedrooms': [2, 3, 3, 4, 3, 4],
    'Location': ['Delhi', 'Noida', 'Delhi', 'Gurgaon', 'Noida', 'Delhi'],
    'Price': [40, 55, 75, 90, 65, 85]
})

# Convert location into numeric columns
df = pd.get_dummies(df, columns=['Location'], dtype=int)

X = df.drop('Price', axis=1)
y = df['Price']

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict price for a new house
new_house = pd.DataFrame([{
    'Area': 1800,
    'Bedrooms': 3,
    'Location_Delhi': 0,
    'Location_Gurgaon': 0,
    'Location_Noida': 1
}])

print("Predicted Price:", round(model.predict(new_house)[0], 2), "Lakhs")
```
